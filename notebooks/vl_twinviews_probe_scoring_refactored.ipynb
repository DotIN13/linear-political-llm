# TwinViews Text Probe Scoring (Refactored)

From `data/twinviews-13k.csv`: left-leaning vs right-leaning sentence pairs.
Interleave them and score each token with a DW-NOMINATE probe.
Visualize token scores as background colors (blue = Dem, red = Rep).

**Refactored**: reuses `probes/` package for probe loading and
`scripts/probes/token_scoring.py` for runtime building, activation capture, and scoring.

```python
import json, os, sys, pickle, random
from collections import defaultdict
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
from IPython.display import HTML, display
from transformers import AutoModelForImageTextToText, AutoProcessor

ROOT = Path.cwd()
if not (ROOT / "results").exists():
    ROOT = Path("/project/jevans/tzhang3/dotty-project/linear-political-llm")
sys.path.insert(0, str(ROOT))

# --- Import shared probe infrastructure ---
from probes import HeadwiseLinearProbe
from probes.base import get_head_module_names
from scripts.probes.token_scoring import (
    build_probe_runtime,
    capture_module_outputs,
    score_from_captured,
    encode_prompts,
    gather_candidate_image_token_ids,
    qwen3_grid_hw_from_path,
    gemma4_grid_hw_from_encoded,
    reconstruct_grid_hw,
    move_to_device,
    select_model_loader,
)
```

```python
# --- Model configuration ---
_MODEL_PATHS = {
    "qwen3-vl": "/project/jevans/tzhang3/models/Qwen3-VL-8B-Instruct",
    "gemma4":   "/project/jevans/tzhang3/models/gemma-4-31B-it",
}
MODEL_FAMILY = "gemma4"  # "qwen3-vl" or "gemma4"
MODEL_PATH = _MODEL_PATHS[MODEL_FAMILY]
PROBE_PREFIX = "textual_ideology"
PROBE_TYPE = "headwise_linear"
TOP_K = 16

print(f"Loading model: {MODEL_PATH}  (family: {MODEL_FAMILY})")
processor = AutoProcessor.from_pretrained(MODEL_PATH)

# Use shared model-loader logic that adapts to model family
model_cls = select_model_loader(MODEL_FAMILY, MODEL_PATH)
if MODEL_FAMILY == "qwen3-vl":
    load_kwargs = {"dtype": torch.bfloat16, "device_map": "auto"}
elif MODEL_FAMILY == "gemma4":
    load_kwargs = {"torch_dtype": torch.bfloat16, "device_map": "auto"}
else:
    load_kwargs = {"torch_dtype": torch.bfloat16, "device_map": "auto"}

model = model_cls.from_pretrained(MODEL_PATH, **load_kwargs).eval()
print("Model loaded.")
```

```python
# Load probe via the shared probes.HeadwiseLinearProbe class
probe = HeadwiseLinearProbe(
    model_path=MODEL_PATH,
    prefix=PROBE_PREFIX,
    mode="text",
    model_family=MODEL_FAMILY,
    data_dir=f"{ROOT}/results/probes",
).load()

print(f"Probe: {probe.scores_.shape[0]}L x {probe.scores_.shape[1]}H")
print(f"  Best head Spearman r: {probe.scores_.max():.4f}")

# Build runtime via shared build_probe_runtime() from token_scoring.py
runtime = build_probe_runtime(
    model=model,
    probe=probe,
    top_k=TOP_K,
    mode="text",
    model_family=MODEL_FAMILY,
)

module_names = sorted(runtime["module_names"])
print(f"  Using top-{TOP_K} heads across {len(module_names)} layers")
print(f"  Normalizer: {runtime['normalizer']}")
```

```python
# Tokenize + score helpers, now delegating to shared infrastructure

def tokenize_text(text):
    """Tokenize raw text directly via processor.tokenizer (no chat template)."""
    enc = processor.tokenizer(text, return_tensors="pt")
    return {k: v.to(model.device) if isinstance(v, torch.Tensor) else v
            for k, v in enc.items()}

def score_tokens(inputs):
    """Capture activations and score using shared capture_module_outputs/score_from_captured."""
    captured = capture_module_outputs(
        model=model,
        encoded=inputs,
        module_names=module_names,
    )
    scores = score_from_captured(captured, runtime)
    return scores.cpu().numpy()[0]

def tokenize_and_score(text):
    """Convenience: tokenize + score, return (tokens, scores)."""
    inputs = tokenize_text(text)
    scores = score_tokens(inputs)
    ids = inputs["input_ids"][0].cpu().numpy()
    toks = [processor.tokenizer.decode([tid]) for tid in ids]
    if len(scores) > len(toks):
        scores = scores[:len(toks)]
    elif len(scores) < len(toks):
        scores = np.pad(scores, (0, len(toks) - len(scores)), constant_values=np.nan)
    return toks, scores

def score_tokens_with_tokenizer(input_ids):
    """Score using inputs dict already on device (e.g. from chat template).
    Returns numpy array of per-token scores."""
    captured = capture_module_outputs(
        model=model,
        encoded=input_ids,
        module_names=module_names,
    )
    scores = score_from_captured(captured, runtime)
    return scores.cpu().numpy()[0]
```

## Token-Level Visualization

Each token is colored: blue (negative = Democratic) -> white -> red (positive = Republican).

```python
def score_to_color(score, vmax=0.5):
    if np.isnan(score):
        return "#e0e0e0"
    t = max(-1., min(1., score / vmax))
    if t <= 0:
        return f"#{int(255*(1+t)):02x}{int(255*(1+t)):02x}ff"
    else:
        return f"#ff{int(255*(1-t)):02x}{int(255*(1-t)):02x}"

def render_html(tokens, scores, vmax=0.5):
    parts = ['<div style="font-family:monospace; line-height:2.4; font-size:13px; max-width:900px; word-wrap:break-word;">']
    for tok, sc in zip(tokens, scores):
        bg = score_to_color(sc, vmax)
        esc = tok.replace("&","&amp;").replace("<","&lt;").replace(">","&gt;")
        parts.append(f'<span style="background:{bg}; padding:2px 1px; margin:0 1px; border-radius:2px;" title="{sc:.4f}">{esc}</span>')
    parts.append('</div>')
    return "".join(parts)

def side_by_side(left_text, right_text, vmax=0.5):
    lt, ls = tokenize_and_score(left_text)
    rt, rs = tokenize_and_score(right_text)
    print(f"  L mean: {np.nanmean(ls):+.4f}    R mean: {np.nanmean(rs):+.4f}")
    return HTML(
        '<table><tr>'
        '<td style="vertical-align:top; padding:6px; width:450px;">'
        f'<b style="color:#5bc0de;">Left</b><br>{render_html(lt, ls, vmax)}</td>'
        '<td style="vertical-align:top; padding:6px; width:450px;">'
        f'<b style="color:#d9534f;">Right</b><br>{render_html(rt, rs, vmax)}</td>'
        '</tr></table>')
```

## Interleaved: L1, R1, L2, R2, ... across many topics

```python
random.seed(42)
sample = tv_df.sample(n=20, random_state=123).reset_index(drop=True)

parts = []
boundaries = []
pos = 0
for _, row in sample.iterrows():
    ltxt = f"LEFT: {row['l']}"
    rtxt = f"RIGHT: {row['r']}"
    lt, _ = tokenize_and_score(ltxt)
    rt, _ = tokenize_and_score(rtxt)
    parts.append(ltxt)
    parts.append(rtxt)
    boundaries.append((pos, "L", row["topic"]))
    pos += len(lt)
    boundaries.append((pos, "R", row["topic"]))
    pos += len(rt)

full = " ".join(parts)
tokens, scores = tokenize_and_score(full)
print(f"Tokens: {len(tokens)}")

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(18, 8), sharex=True,
                               gridspec_kw={"height_ratios": [3, 1]})
x = np.arange(len(scores))
ax1.fill_between(x, 0, scores, where=scores > 0, color="#d9534f", alpha=0.6, step="mid")
ax1.fill_between(x, 0, scores, where=scores < 0, color="#5bc0de", alpha=0.6, step="mid")
ax1.axhline(0, color="black", linewidth=0.5)
ax1.set_ylabel("DW-NOMINATE score")
ax1.set_title("Per-Token Probe Score Across Interleaved Left/Right Sentences")

for i in range(len(boundaries)):
    bpos, bside, btop = boundaries[i]
    next_b = boundaries[i+1][0] if i+1 < len(boundaries) else len(scores)
    color = "#5bc0de" if bside == "L" else "#d9534f"
    ax1.axvspan(bpos, min(next_b, len(scores)), alpha=0.06, color=color)

diffs = []
for i in range(0, len(boundaries), 2):
    if i+1 < len(boundaries):
        ls = scores[boundaries[i][0]:boundaries[i+1][0]]
        rs_end = boundaries[i+2][0] if i+2 < len(boundaries) else len(scores)
        rs = scores[boundaries[i+1][0]:rs_end]
        d = np.nanmean(ls) - np.nanmean(rs)
        diffs.append(d)
        mid = (boundaries[i][0] + boundaries[i+1][0]) / 2
        ax2.scatter(mid, d, c="#333", s=25, zorder=5)

ax2.axhline(0, color="black", linewidth=0.5)
ax2.set_ylabel("mean(L) - mean(R)")
ax2.set_xlabel("Token position")
ax2.set_title("Per-Pair Difference (negative = L more Dem-leaning)")
plt.tight_layout()
plt.show()

print(f"Overall mean: {np.nanmean(scores):+.4f}")
print(f"L - R (mean of per-pair diffs): {np.mean(diffs):+.4f}")
```

## Batch: Distribution of Sentence-Level Means

```python
batch = tv_df.sample(n=100, random_state=42)
rows = []
for _, row in batch.iterrows():
    _, ls = tokenize_and_score(row["l"])
    _, rs = tokenize_and_score(row["r"])
    rows.append({"topic": row["topic"], "mean_score": np.nanmean(ls), "side": "Left"})
    rows.append({"topic": row["topic"], "mean_score": np.nanmean(rs), "side": "Right"})

comp = pd.DataFrame(rows)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for side, c, ax in [("Left", "#5bc0de", axes[0]), ("Right", "#d9534f", axes[0])]:
    v = comp[comp["side"] == side]["mean_score"].dropna()
    ax.hist(v, bins=25, alpha=0.5, color=c, label=f"{side} (u={v.mean():+.3f})")
axes[0].axvline(0, color="black", linestyle="--", linewidth=0.8)
axes[0].set_title("Mean Token Score by Political Leaning")
axes[0].set_xlabel("Mean DW-NOMINATE probe score")
axes[0].legend()

# Paired difference
paired = comp.pivot_table(index="topic", columns="side", values="mean_score", aggfunc="mean").dropna()
paired["diff"] = paired["Left"] - paired["Right"]
axes[1].hist(paired["diff"], bins=20, color="#888888", edgecolor="white")
axes[1].axvline(0, color="black", linestyle="--", linewidth=0.8)
axes[1].set_title(f"Paired Difference (Left - Right)\nmean={paired['diff'].mean():+.4f}")
axes[1].set_xlabel("mean(Left) - mean(Right)")
plt.tight_layout()
plt.show()

from scipy import stats
if len(paired) > 2:
    t_stat, t_p = stats.ttest_1samp(paired["diff"], 0)
    print(f"Paired t-test: t={t_stat:.3f}, p={t_p:.3g}")
print(f"Left mean:  {paired['Left'].mean():+.4f}")
print(f"Right mean: {paired['Right'].mean():+.4f}")
```

## Image Canvas: Interleaved L/R Text Rendered as an Image

Render left/right sentences onto a PIL image canvas line by line,
then feed the image through the VLM and score the tokens.

**Uses** `encode_prompts` from `token_scoring.py` for chat template application,
`gather_candidate_image_token_ids` for image token detection, and
`qwen3_grid_hw_from_path` for grid dimension computation.

```python
from PIL import Image, ImageDraw, ImageFont

# Try to load a font, fall back to default
font = ImageFont.load_default(32)

# ---- Side-mask configuration ----
# Main canvas colours — freely changeable (used only for display)
MAIN_COLORS = {"L": "#000000", "R": "#000000"}
# Mask canvas colours — fixed (used for decoding side identity; do NOT change)
MASK_COLORS = {"L": "#0000ff", "R": "#ff0000"}

def decode_side_mask(mask_img):
    """Convert an RGB mask image to a 2D side-label array.

    Left  text is drawn in blue  (#0000ff) -> label 0
    Right text is drawn in red   (#ff0000) -> label 1
    Background (white or other)  -> label 255

    Works on any resolution; for best results downscale the RGB mask
    to the target grid size with LANCZOS first, then call this.
    """
    arr = np.array(mask_img).astype(np.int16)
    is_left  = (arr[:,:,2] > arr[:,:,0] + 10) & (arr[:,:,2] > arr[:,:,1] + 10)
    is_right = (arr[:,:,0] > arr[:,:,2] + 10) & (arr[:,:,0] > arr[:,:,1] + 10)
    result = np.full(arr.shape[:2], 255, dtype=np.uint8)
    result[is_left]  = 0
    result[is_right] = 1
    return result

# Build interleaved left/right lines (side label instead of colour string)
few = tv_df.sample(n=4, random_state=42).reset_index(drop=True)
lines = []
for _, row in few.iterrows():
    lines.append((f"{row['l']}", "L"))
    lines.append((f"{row['r']}", "R"))

# Compute canvas size with word-wrapping
line_spacing = 4
sent_gap = 16
max_width = 900
line_h = font.getbbox("Ag")[3] - font.getbbox("Ag")[1] + line_spacing

wrapped_lines = []
wrapped_sides = []
wrapped_sent_idx = []
for si, (text, side) in enumerate(lines):
    words = text.split()
    cur = ""
    for w in words:
        test = cur + (" " if cur else "") + w
        if font.getbbox(test)[2] - font.getbbox(test)[0] > max_width - 20:
            wrapped_lines.append(cur)
            wrapped_sides.append(side)
            wrapped_sent_idx.append(si)
            cur = w
        else:
            cur = test
    if cur:
        wrapped_lines.append(cur)
        wrapped_sides.append(side)
        wrapped_sent_idx.append(si)

sent_gaps = sum(1 for i in range(1, len(wrapped_sent_idx)) if wrapped_sent_idx[i] != wrapped_sent_idx[i-1])
canvas_h = len(wrapped_lines) * line_h + sent_gaps * sent_gap + 20
canvas = Image.new("RGB", (max_width, canvas_h), "white")
draw = ImageDraw.Draw(canvas)

# Side mask: drawn in fixed red/blue, identical layout to main canvas
side_mask = Image.new("RGB", (max_width, canvas_h), "white")
draw_mask = ImageDraw.Draw(side_mask)

y = 10
for i, (text, side) in enumerate(zip(wrapped_lines, wrapped_sides)):
    if i > 0 and wrapped_sent_idx[i] != wrapped_sent_idx[i-1]:
        y += sent_gap  # gap between sentences
    draw.text((10, y), text, fill=MAIN_COLORS[side], font=font)
    draw_mask.text((10, y), text, fill=MASK_COLORS[side], font=font)
    y += line_h

display(canvas.resize((max_width//2, canvas_h//2)))
print(f"Canvas: {canvas.size}, {len(wrapped_lines)} lines")
```

```python
# Feed image through VLM and score — using shared infrastructure
# 1) Build messages & encode via shared encode_prompts()
msgs = [{"role": "user", "content": [
    {"type": "image", "image": canvas},
    {"type": "text", "text": "What political perspectives are expressed in this image?"},
]}]
enc = encode_prompts(processor, [msgs])
inputs = move_to_device(enc, model)

# 2) Find image tokens via shared gather_candidate_image_token_ids()
image_token_ids = gather_candidate_image_token_ids(processor.tokenizer)
image_token_array = np.asarray(sorted(image_token_ids), dtype=np.int64)
input_ids_np = enc["input_ids"][0].cpu().numpy()
is_image = np.isin(input_ids_np, image_token_array)

# 3) Compute vision grid size via shared reconstruct_grid_hw() — handles Qwen3‑VL (temp‑file) and Gemma4
if MODEL_FAMILY == "qwen3-vl":
    import tempfile
    tmp_path = None
    try:
        with tempfile.NamedTemporaryFile(suffix=".png", delete=False) as tmp:
            canvas.save(tmp.name)
            tmp_path = tmp.name
        gh, gw = reconstruct_grid_hw(MODEL_FAMILY, tmp_path, enc, processor)
    finally:
        if tmp_path:
            os.unlink(tmp_path)
else:
    gh, gw = reconstruct_grid_hw(MODEL_FAMILY, "", enc, processor)


# ---- Downscale RGB side_mask to vision grid, then decode ----
mask_small = side_mask.resize((gw, gh), Image.LANCZOS)
mask_grid = decode_side_mask(mask_small)  # (gh, gw): 0=L, 1=R, 255=bg
mask_grid
```

```python
from matplotlib.patches import Patch

# 4) Score all tokens via shared capture/output — note: use the already-device inputs
all_scores = score_tokens_with_tokenizer(inputs)
img_scores = all_scores[is_image]
n_img = int(is_image.sum())
print(f"Image tokens: {n_img} (grid {gh}x{gw})")
print(f"Image mean: {img_scores.mean():+.4f}")

# ---- Text token sequence plot ----
parts = []
boundaries = []
pos = 0

for _, row in few.iterrows():
    ltxt = f"LEFT: {row['l']}"
    rtxt = f"RIGHT: {row['r']}"

    lt, _ = tokenize_and_score(ltxt)
    rt, _ = tokenize_and_score(rtxt)

    parts.append(ltxt)
    parts.append(rtxt)

    boundaries.append((pos, "L", row.get("topic", "")))
    pos += len(lt)

    boundaries.append((pos, "R", row.get("topic", "")))
    pos += len(rt)

full = " ".join(parts)
tokens, scores = tokenize_and_score(full)
print(f"Tokens: {len(tokens)}")

fig, ax = plt.subplots(figsize=(16, 5))

TITLE_SIZE = 12
LABEL_SIZE = 10
TICK_SIZE = 10
LEGEND_SIZE = 10

x = np.arange(len(scores))

ax.fill_between(
    x, 0, scores,
    where=scores > 0,
    color="#d9534f",
    alpha=0.5,
    step="mid"
)

ax.fill_between(
    x, 0, scores,
    where=scores < 0,
    color="#5bc0de",
    alpha=0.5,
    step="mid"
)

for i, (bpos, side, topic) in enumerate(boundaries):
    next_b = boundaries[i + 1][0] if i + 1 < len(boundaries) else len(scores)

    if side == "L":
        color = "#5bc0de"
    elif side == "R":
        color = "#d9534f"
    else:
        color = "#cccccc"

    ax.axvspan(
        bpos,
        min(next_b, len(scores)),
        alpha=0.12,
        color=color
    )

for bpos, side, topic in boundaries[1:]:
    ax.axvline(
        bpos,
        color="#888888",
        linewidth=0.3,
        linestyle=":"
    )

legend_elements = [
    Patch(facecolor="#5bc0de", alpha=0.3, label="Left sentence regions"),
    Patch(facecolor="#d9534f", alpha=0.3, label="Right sentence regions"),
]

ax.legend(
    handles=legend_elements,
    loc="upper right",
    fontsize=LEGEND_SIZE
)

ax.axhline(0, color="black", linewidth=0.5)

ax.set_xlabel(
    "Token index across interleaved Left/Right sentences",
    fontsize=LABEL_SIZE
)
ax.set_ylabel(
    "DW-NOMINATE score",
    fontsize=LABEL_SIZE
)
ax.set_title(
    "Text Tokens",
    fontsize=TITLE_SIZE
)

ax.tick_params(axis="both", labelsize=TICK_SIZE)

plt.tight_layout()
plt.show()

l_scores = []
r_scores = []

for i, (bpos, side, topic) in enumerate(boundaries):
    next_b = boundaries[i + 1][0] if i + 1 < len(boundaries) else len(scores)
    span_scores = scores[bpos:next_b]

    if side == "L":
        l_scores.extend(span_scores)
    elif side == "R":
        r_scores.extend(span_scores)

print(f"Overall mean: {np.nanmean(scores):+.4f}")
print(
    f"L sentence tokens ({len(l_scores)}): mean={np.nanmean(l_scores):+.4f}"
    if l_scores else
    "L sentence tokens: none"
)
print(
    f"R sentence tokens ({len(r_scores)}): mean={np.nanmean(r_scores):+.4f}"
    if r_scores else
    "R sentence tokens: none"
)
```

## Multi-Sample: Image Canvas vs Direct Text Correlation

Run N twinview pairs through BOTH pipelines:
1. **Image canvas**: render L/R text as an image, score vision tokens, extract per-sentence region means
2. **Direct text**: tokenize each sentence separately, compute per-sentence mean

Then compare: sentence-level and token-level correlation.

```python
N_ITERS = 50
PAIRS_PER_ITER = 4
random.seed(42)

records = []

sharp = tv_df['l'].apply(len) + tv_df['r'].apply(len)
keep = sharp[sharp < 400]
tv_filt = tv_df.loc[keep.index]

for it in range(N_ITERS):
    sample_pairs = tv_filt.sample(n=PAIRS_PER_ITER, random_state=it).reset_index(drop=True)

    # ---- Build interleaved canvas ----
    lines = []
    for _, row in sample_pairs.iterrows():
        lines.append((f"L: {row['l']}", "L"))
        lines.append((f"R: {row['r']}", "R"))

    wrapped_lines = []
    wrapped_sides = []
    wrapped_pair_idx = []
    wrapped_side = []
    pi = 0
    for text, side_ in lines:
        words = text.split()
        cur = ""
        for w in words:
            test = cur + (" " if cur else "") + w
            if font.getbbox(test)[2] - font.getbbox(test)[0] > max_width - 20:
                wrapped_lines.append(cur)
                wrapped_sides.append(side_)
                wrapped_pair_idx.append(pi)
                wrapped_side.append(side_)
                cur = w
            else:
                cur = test
        if cur:
            wrapped_lines.append(cur)
            wrapped_sides.append(side_)
            wrapped_pair_idx.append(pi)
            wrapped_side.append(side_)
        if side_ == "R":
            pi += 1

    sent_gaps = sum(1 for i in range(1, len(wrapped_pair_idx))
                   if (wrapped_pair_idx[i], wrapped_side[i]) != (wrapped_pair_idx[i-1], wrapped_side[i-1]))
    canvas_h = len(wrapped_lines) * line_h + sent_gaps * sent_gap + 20
    canvas = Image.new("RGB", (max_width, canvas_h), "white")
    draw = ImageDraw.Draw(canvas)

    # Side mask: drawn in fixed red/blue, identical layout
    side_mask = Image.new("RGB", (max_width, canvas_h), "white")
    draw_mask = ImageDraw.Draw(side_mask)

    y = 10
    for i, (text, side) in enumerate(zip(wrapped_lines, wrapped_sides)):
        if i > 0 and (wrapped_pair_idx[i], wrapped_side[i]) != (wrapped_pair_idx[i-1], wrapped_side[i-1]):
            y += sent_gap  # gap between sentences
        draw.text((10, y), text, fill=MAIN_COLORS[side], font=font)
        draw_mask.text((10, y), text, fill=MASK_COLORS[side], font=font)
        y += line_h

    # ---- Score via VLM (shared infra) ----
    msgs = [{"role": "user", "content": [
        {"type": "image", "image": canvas},
        {"type": "text", "text": "What political perspectives are expressed in this image?"},
    ]}]
    enc = encode_prompts(processor, [msgs])
    inputs = move_to_device(enc, model)

    input_ids_np = enc["input_ids"][0].cpu().numpy()
    is_image = np.isin(input_ids_np, image_token_array)

    if MODEL_FAMILY == "qwen3-vl":
        import tempfile
        tmp_path = None
        try:
            with tempfile.NamedTemporaryFile(suffix=".png", delete=False) as tmp:
                canvas.save(tmp.name)
                tmp_path = tmp.name
            gh, gw = reconstruct_grid_hw(MODEL_FAMILY, tmp_path, enc, processor)
        finally:
            if tmp_path:
                os.unlink(tmp_path)
    else:
        gh, gw = reconstruct_grid_hw(MODEL_FAMILY, "", enc, processor)

    # Downscale RGB side_mask to vision grid, then decode
    mask_small = side_mask.resize((gw, gh), Image.LANCZOS)
    mask_grid = decode_side_mask(mask_small)

    all_scores = score_tokens_with_tokenizer(inputs)
    img_scores = all_scores[is_image]
    img_grid = img_scores.reshape(gh, gw)

    # ---- Map grid rows to pairs/sentences ----
    grid_row_height = canvas_h / gh
    row_info = [(-1, "") for _ in range(gh)]
    line_y = []
    y = 10
    for i in range(len(wrapped_lines)):
        if i > 0 and (wrapped_pair_idx[i], wrapped_side[i]) != (wrapped_pair_idx[i-1], wrapped_side[i-1]):
            y += sent_gap
        line_y.append(y)
        y += line_h
    for li in range(len(wrapped_lines)):
        y_start = line_y[li]
        y_end = y_start + line_h - line_spacing
        row_start = int(y_start / grid_row_height)
        row_end = min(int(y_end / grid_row_height) + 1, gh)
        for r in range(row_start, row_end):
            if 0 <= r < gh:
                row_info[r] = (wrapped_pair_idx[li], wrapped_side[li])

    # ---- Extract per-sentence data ----
    for pp, (_, row) in enumerate(sample_pairs.iterrows()):
        for side, key in [("L", "l"), ("R", "r")]:
            toks, tscores = tokenize_and_score(row[key])
            txt_mean = np.nanmean(tscores)

            img_rows = [r for r in range(gh) if row_info[r] == (pp, side)]
            if img_rows:
                img_region = img_grid[img_rows, :]
                img_mean = float(img_region.mean())
                img_flat = img_region.flatten().tolist()
            else:
                img_mean = float('nan')
                img_flat = []

            records.append({
                "iter": it, "pair_idx": pp, "side": side,
                "txt_scores": tscores.tolist() if len(tscores) > 0 else [],
                "txt_mean": txt_mean,
                "img_n_cells": len(img_flat),
                "img_mean": img_mean,
                "img_flat": img_flat,
            })

    if (it + 1) % 10 == 0:
        print(f"  iter {it+1}/{N_ITERS}  ({len(records)} sentences so far)")

df_records = pd.DataFrame(records)
valid = df_records.dropna(subset=["txt_mean", "img_mean"])
print(f"Done: {len(df_records)} total sentences, {len(valid)} valid ({N_ITERS} iters x {PAIRS_PER_ITER} pairs x 2)")
print(f"Image cells/sent: min={df_records['img_n_cells'].min()}, max={df_records['img_n_cells'].max()}, median={df_records['img_n_cells'].median():.0f}")
print(f"Text tokens/sent:  min={df_records['txt_scores'].apply(len).min()}, max={df_records['txt_scores'].apply(len).max()}")
```

## Sentence-Level Correlation

For each L or R sentence, compute the mean text-token score and the mean image-region score.
Scatter and correlate.

```python
from scipy import stats

valid = df_records.dropna(subset=["txt_mean", "img_mean"])

plt.rcParams.update({
    "font.family": "Arial, sans-serif",
    "font.size": 8,
    "axes.linewidth": 0.6,
    "axes.labelsize": 8,
    "axes.titlesize": 9,
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "legend.fontsize": 7,
    "xtick.major.width": 0.6,
    "ytick.major.width": 0.6,
    "xtick.major.size": 3,
    "ytick.major.size": 3,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

fig, ax = plt.subplots(figsize=(5, 4))

palette = {
    "L": "#4E79A7",
    "R": "#E15759",
}
markers = {
    "L": "o",
    "R": "s",
}

for side in ["L", "R"]:
    sub = valid[valid["side"] == side]
    ax.scatter(
        sub["txt_mean"],
        sub["img_mean"],
        c=palette[side],
        marker=markers[side],
        s=22,
        edgecolors="white",
        linewidth=0.4,
        alpha=0.85,
        label=side,
        zorder=3,
    )

ax.axhline(0, color="0.75", linewidth=0.6, linestyle="--", zorder=1)
ax.axvline(0, color="0.75", linewidth=0.6, linestyle="--", zorder=1)

r_all, p_all = stats.pearsonr(valid["txt_mean"], valid["img_mean"])
rho_all, p_rho_all = stats.spearmanr(valid["txt_mean"], valid["img_mean"])

label_str = (
    f"Pearson r = {r_all:.2f}, p = {p_all:.2g}\n"
    f"Spearman ρ = {rho_all:.2f}, p = {p_rho_all:.2g}"
)

ax.text(
    0.04,
    0.96,
    label_str,
    transform=ax.transAxes,
    ha="left",
    va="top",
    fontsize=7,
)

ax.set_xlabel("Mean text token score")
ax.set_ylabel("Mean image-region score")
ax.set_title(f"Text-image score correspondence, n = {len(valid)}", pad=6)

leg = ax.legend(
    frameon=False,
    loc="lower right",
    handletextpad=0.3,
    borderaxespad=0.3,
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(True, color="0.9", linewidth=0.4, zorder=0)
plt.tight_layout()
# plt.savefig("text_image_correlation_nature_style_refactored.pdf", bbox_inches="tight")
# plt.savefig("text_image_correlation_nature_style_refactored.png", dpi=600, bbox_inches="tight")
plt.show()

print(
    f"All sentences (n={len(valid)}): "
    f"Pearson r={r_all:.4f} p={p_all:.4f}  "
    f"Spearman rho={rho_all:.4f} p={p_rho_all:.4f}"
)

for side in ["L", "R"]:
    sub = valid[valid.side == side]
    r_side, p_side = stats.pearsonr(sub["txt_mean"], sub["img_mean"])
    rho_side, p_rho_side = stats.spearmanr(sub["txt_mean"], sub["img_mean"])

    print(
        f"{side} only (n={len(sub)}): "
        f"Pearson r={r_side:.4f} p={p_side:.4f}  "
        f"Spearman rho={rho_side:.4f} p={p_rho_side:.4f}"
    )
```
